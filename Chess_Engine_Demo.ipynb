# Chess Engine Demonstration
This notebook demonstrates the capabilities of the custom chess engine. The engine is written in pure Python and supports all standard chess rules including castling, en passant, promotion, and checkmate detection.

## 1. Board Representation & Initial State
The board is represented using an 8x8 grid. We can easily initialize it and view the starting position.

```python
from chess.board import Board
from engine.agent import ChessAgent
from engine.evaluation import Evaluator

board = Board()
print("Current Turn:", board.current_turn)
board.print_board()
```

## 2. Legal Move Generation & Perft Validation
The engine correctly generates all legal moves (20 for the starting position).

```python
moves = []
for r in range(8):
    for c in range(8):
        p = board.position[r][c]
        if p and p.color == board.current_turn:
            moves.extend(board.generate_legal_moves(r, c))
print(f"Number of legal moves for {board.current_turn}: {len(moves)}")
```

## 3. Make and Undo Moves
The engine supports making and undoing moves. This is crucial for the Minimax search algorithm.

```python
# Play e2-e4
e4_move = next((m for m in moves if m.start == (6, 4) and m.end == (4, 4)))
board.make_move(e4_move)
board.current_turn = "black"
print("After e2-e4:")
board.print_board()

# Undo the move
board.current_turn = "white"
board.undo_move()
print("\nAfter undo:")
board.print_board()
```

## 4. Special Rules: Castling, En Passant, Promotion
The move generator natively detects and handles all special chess rules correctly.

```python
print("Initial Castling Rights:", board.castling_rights)
print("Initial En Passant Square:", board.en_passant_square)
```

## 5. Evaluation and AI Minimax Search
The engine uses a Minimax algorithm with Alpha-Beta pruning, MVV-LVA move ordering, and Quiescence search to find the best move. Let's ask the AI to find a Mate in 1.

```python
# Setup Scholar's Mate scenario
board.make_move(next((m for m in board.generate_legal_moves(6, 4) if m.end == (4, 4)))) # e4
board.current_turn = "black"
board.make_move(next((m for m in board.generate_legal_moves(1, 4) if m.end == (3, 4)))) # e5
board.current_turn = "white"
board.make_move(next((m for m in board.generate_legal_moves(7, 5) if m.end == (4, 2)))) # Bc4
board.current_turn = "black"
board.make_move(next((m for m in board.generate_legal_moves(0, 1) if m.end == (2, 2)))) # Nc6
board.current_turn = "white"
board.make_move(next((m for m in board.generate_legal_moves(7, 3) if m.end == (3, 7)))) # Qh5
board.current_turn = "black"
board.make_move(next((m for m in board.generate_legal_moves(0, 6) if m.end == (2, 5)))) # Nf6
board.current_turn = "white"

print("Board before AI Move:")
board.print_board()

import time
agent = ChessAgent(depth=2)
start_time = time.time()
best_move = agent.get_move(board)
search_time = time.time() - start_time

print(f"\nAI found best move in {search_time:.3f} seconds!")
print(f"Best move: {best_move.start} -> {best_move.end}")
```

## 6. Capabilities and Status
- **Board & Move Generation**: Fully Implemented. Includes Castling, Promotion, En Passant.
- **Search Algorithm**: Minimax with Alpha-Beta Pruning.
- **Move Ordering**: MVV-LVA (Most Valuable Victim - Least Valuable Attacker) heuristics to optimize Alpha-Beta cutoffs.
- **Quiescence Search**: Extends search past the set depth to ensure tactical stability (prevents horizon effect).
- **Validation**: Validated with Perft tests to match standard chess move generation counts.
